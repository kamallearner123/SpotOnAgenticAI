```python
# 3. Memory : To reduce the traffic b/w User and LLM service
# 2. Decision making
# 1. Listen


memory = []
while True:
    user = input("User:") # Asking user to enter input "Country"

    # Add logic to check the Q, is already there in memory
    # If it is there in memory, print A and continue
    
    if user == "exit":
        break
    elif "Python" in user:
        # Go and ask LLM / Send query to ChatGTP
        answer = "Python langauge is very import of AI/ML"
    elif "India" in user:
        answer = "India is a great country"
    else: # If there is not Answer, continue
        continue
    # Output
    print("Agent: ", answer)
    memory.append({"Q":user, "A":answer})
    
print(memory)
```
