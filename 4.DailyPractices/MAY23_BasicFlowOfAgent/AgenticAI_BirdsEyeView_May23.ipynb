```python
# 2. Decision making
# 1. Listen

while True:
    user = input("User:")
    if user == "exit":
        break
    elif "Python" in user:
        answer = "Python langauge is very import of AI/ML"
    elif "India" in user:
        answer = "India is a great country"
    # Output
    print("Agent: ", answer)
```

```python
# 3. Memory
# 2. Decision making
# 1. Listen


memory = []
while True:
    user = input("User:")
    if user == "exit":
        break
    elif "Python" in user:
        answer = "Python langauge is very import of AI/ML"
    elif "India" in user:
        answer = "India is a great country"
    # Output
    print("Agent: ", answer)
    memory.append({"Q":user, "A":answer})
print(memory)
```

```python
# 4. Persistance (logs)
# 3. Memory
# 2. Decision making
# 1. Listen

# DB, Text (json)

#importing json module
import json

memory = []
while True:
    user = input("User:")
    if user == "exit":
        break
    elif "Python" in user:
        answer = "Python langauge is very import of AI/ML"
    elif "India" in user:
        answer = "India is a great country"
    # Output
    print("Agent: ", answer)
    memory.append({"Q":user, "A":answer})
print(memory)
with open("Memory.json", "w") as fd:
    json.dump(memory, fd)
```
